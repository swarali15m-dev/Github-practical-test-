Exploratory Data Analysis of House Prices
Project Overview
The objective of this project is to perform Exploratory Data Analysis (EDA) on the Ames Housing Dataset.

The project focuses on understanding different factors that affect house sale prices. We will analyse numerical and categorical data, calculate statistical measures, identify missing values and outliers, and create visualizations.

Dataset Description
The dataset contains information about houses, including their area, quality, neighborhood, year of construction, and sale price.

Objectives
Understand the distribution of house prices.
Identify features correlated with SalePrice.
Analyse the effect of overall quality on house prices.
Compare house prices across neighborhoods.
Detect outliers and missing values.
Create visualizations and a final dashboard.

```python

# EDA PROJECT: HOUSE PRICE ANALYSIS

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Load dataset automatically
url = "https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv"
df = pd.read_csv(url)

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
display(df.isnull().sum().sort_values(ascending=False).head(20))

print("\nDescriptive Statistics:")
display(df.describe())
```

```python
# Basic information
print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
display(df.isnull().sum().sort_values(ascending=False).head(15))

# Numerical statistics
print("\nDescriptive Statistics:")
display(df.describe())

# SalePrice analysis
if 'SalePrice' in df.columns:
    print("\nSalePrice Mean:", df['SalePrice'].mean())
    print("SalePrice Median:", df['SalePrice'].median())
    print("SalePrice Standard Deviation:", df['SalePrice'].std())

    import matplotlib.pyplot as plt
    import seaborn as sns

    plt.figure(figsize=(8, 4))
    sns.histplot(df['SalePrice'].dropna(), kde=True)
    plt.title("Distribution of House Prices")
    plt.xlabel("Sale Price")
    plt.show()
```

```python
# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

# Correlation with SalePrice
if 'SalePrice' in numeric_df.columns:
    correlation = numeric_df.corr()['SalePrice'].sort_values(ascending=False)

    print("Top 10 features correlated with SalePrice:")
    display(correlation.head(11))

# Correlation heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), cmap='coolwarm', center=0)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Scatter plot: Total Rooms vs Median House Value
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='total_rooms', y='median_house_value')
plt.title("Total Rooms vs Median House Value")
plt.show()

# Median Income vs Median House Value
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='median_income', y='median_house_value')
plt.title("Median Income vs Median House Value")
plt.show()

# Ocean Proximity vs Median House Value
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x='ocean_proximity', y='median_house_value')
plt.title("Ocean Proximity vs Median House Value")
plt.show()
```

```python
# Average house price by ocean proximity
ocean_prices = df.groupby('ocean_proximity')['median_house_value'].mean()
ocean_prices = ocean_prices.sort_values(ascending=False)

print("Average House Value by Ocean Proximity:")
display(ocean_prices)

# Plot the average prices
plt.figure(figsize=(10, 5))
ocean_prices.plot(kind='bar', color='skyblue')
plt.title("Average House Value by Ocean Proximity")
plt.xlabel("Ocean Proximity")
plt.ylabel("Average Median House Value")
plt.xticks(rotation=45)
plt.show()
```

```python
# Detect outliers using IQR method
Q1 = df['median_house_value'].quantile(0.25)
Q3 = df['median_house_value'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['median_house_value'] < lower_limit) |
              (df['median_house_value'] > upper_limit)]

print("Number of Outliers:", len(outliers))

# Boxplot
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['median_house_value'])
plt.title("Outlier Detection in House Prices")
plt.show()
```

```python
# Check missing values
missing_values = df.isnull().sum()

# Display columns with missing values
missing_values = missing_values[missing_values > 0].sort_values(ascending=False)

print("Missing Values in Dataset:")
display(missing_values)

# Visualize missing values
plt.figure(figsize=(12, 5))
missing_values.head(15).plot(kind='bar', color='orange')
plt.title("Top 15 Columns with Missing Values")
plt.xlabel("Columns")
plt.ylabel("Missing Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Check duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

# Fill missing numerical values with median
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values with None
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna('None')

print("\nMissing values after cleaning:")
print(df.isnull().sum().sum())

print("\nData cleaning completed!")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. House Price Distribution
sns.histplot(df['median_house_value'], kde=True, ax=axes[0, 0])
axes[0, 0].set_title("House Price Distribution")

# 2. Average Price by Ocean Proximity
ocean_vals = df.groupby('ocean_proximity')['median_house_value'].mean().sort_values(ascending=False)
ocean_vals.plot(kind='bar', ax=axes[0, 1])
axes[0, 1].set_title("Average Price by Ocean Proximity")
axes[0, 1].tick_params(axis='x', rotation=45)

# 3. Total Rooms vs Median House Value
sns.scatterplot(data=df, x='total_rooms', y='median_house_value',
                hue='ocean_proximity', ax=axes[1, 0])
axes[1, 0].set_title("Total Rooms vs Median House Value")
axes[1, 0].legend(title="Proximity", fontsize=7)

# 4. Average Price by Housing Median Age
age_prices = df.groupby('housing_median_age')['median_house_value'].mean()
age_prices.plot(ax=axes[1, 1])
axes[1, 1].set_title("Average Price by Housing Median Age")

plt.tight_layout()
plt.savefig("eda_dashboard.png")
plt.show()

print("Dashboard created successfully!")
```
